```python
# ===================================================
# INSTALL
# ===================================================

!pip -q install gradio

# ===================================================
# IMPORT
# ===================================================

import gradio as gr
import base64

# ===================================================
# IMAGE
# Colab 왼쪽에 park.png 업로드
# ===================================================

IMAGE_PATH="park.png"

with open(IMAGE_PATH,"rb") as f:
    image_base64=base64.b64encode(f.read()).decode()

image_data=f"data:image/png;base64,{image_base64}"

# ===================================================
# FUNCTIONS
# ===================================================

def save_structure(intro,body,conclusion):

    structure=f"""
Introduction
{intro}

Body
{body}

Conclusion
{conclusion}
"""

    return structure,structure,"✅ Structure saved successfully!"


def save_structure_page3(text):

    return text,text,"✅ Structure updated successfully!"


def count_words(text):

    if not text:
        return "Word Count: 0"

    return f"Word Count: {len(text.split())}"


def submit_essay(text):

    return "✅ Essay submitted successfully!"


def create_final(essay):

    html=f"""

<div style="
width:850px;
margin:auto;
padding:40px;
border:2px solid black;
background:white;
font-family:Arial;
">

<div style="
display:flex;
justify-content:space-between;
align-items:flex-start;
">

<div>

<h1>
My Small Actions for the Environment
</h1>

</div>

<div>

<img src="{image_data}"
width="250">

</div>

</div>

<hr>

<h2>My Essay</h2>

<div style="
line-height:2;
font-size:18px;
padding:20px;
border:1px solid lightgray;
min-height:400px;
">

{essay}

</div>

</div>

"""

    return html


# ===================================================
# APP
# ===================================================

with gr.Blocks(title="Writing Practice App") as demo:

    structure_state=gr.State("")

    gr.Markdown(
    "# 🌎 Writing Practice App"
    )


# ===================================================
# PAGE1
# ===================================================

    with gr.Tab("Page1 Picture"):

        gr.Markdown(
        "## Observe the picture carefully"
        )

        gr.Image(
            value=IMAGE_PATH,
            show_label=False
        )

        gr.Markdown("""

### Think

• What do you see?
• What is happening?
• What can people do for nature?

""")


# ===================================================
# PAGE2
# ===================================================

    with gr.Tab("Page2 Structure"):

        gr.Markdown("""

# Essay Topic

## My Small Actions for the Environment

Write your Introduction - Body - Conclusion.

""")

        intro=gr.Textbox(
            label="Introduction"
        )

        body=gr.Textbox(
            label="Body"
        )

        conclusion=gr.Textbox(
            label="Conclusion"
        )

        save_btn=gr.Button(
            "Save Structure"
        )

        page2_message=gr.Textbox(
            label="",
            interactive=False
        )


# ===================================================
# PAGE3
# ===================================================

    with gr.Tab("Page3 Content Support"):

        gr.Markdown(
        "## Compare with Content Support and revise"
        )

        with gr.Row():

            my_structure=gr.Textbox(
                label="My Structure",
                lines=15
            )

            with gr.Column():

                gr.Markdown("""

# 💡 Content Support

## Introduction

### Think about:

• Why is the environment important?
• Why should we protect nature?
• How does a clean Earth help us?

""")

                with gr.Accordion(
                    "Example Ideas",
                    open=False
                ):

                    gr.Markdown("""

• The Earth is our home
• Clean air and water are important
• We should take care of nature

""")


                gr.Markdown("""

## Body

### Think about:

• What can people do to help the environment?
• What can you do at home or school?
• How can small actions help?

""")

                with gr.Accordion(
                    "Example Ideas ",
                    open=False
                ):

                    gr.Markdown("""

• Turn off lights when not using them
• Save water when brushing teeth
• Recycle paper and bottles
• Use less plastic
• Plant trees
• Pick up trash

""")


                gr.Markdown("""

## Conclusion

### Think about:

• Why do small actions matter?
• What happens if everyone helps?
• What is your hope for the future?

""")

                with gr.Accordion(
                    "Example Ideas  ",
                    open=False
                ):

                    gr.Markdown("""

• Small actions can make a big difference
• Everyone can help the Earth
• We can make the world cleaner

""")


        save_page3=gr.Button(
            "Save Revised Structure"
        )

        page3_message=gr.Textbox(
            label="",
            interactive=False
        )


# ===================================================
# PAGE4
# ===================================================

    with gr.Tab("Page4 Writing"):

        gr.Markdown(
        "## Write your essay"
        )

        with gr.Row():

            plan=gr.Textbox(
                label="Essay Plan",
                lines=15
            )

            essay=gr.Textbox(
                label="Write Here",
                lines=15
            )


        word_counter=gr.Textbox(
            label="Word Count"
        )

        submit_btn=gr.Button(
            "Submit Essay"
        )

        submit_message=gr.Textbox(
            label="",
            interactive=False
        )


# ===================================================
# PAGE5
# ===================================================

    with gr.Tab("Final Printable Page"):

        gr.Markdown(
        "## Final Submission"
        )

        final_output=gr.HTML()

        generate=gr.Button(
            "Generate Final Page"
        )

        gr.Markdown("""

### Print

Windows → Ctrl + P

Mac → Cmd + P

""")


# ===================================================
# EVENTS
# ===================================================

    save_btn.click(
        fn=save_structure,
        inputs=[intro,body,conclusion],
        outputs=[
            my_structure,
            plan,
            page2_message
        ]
    )


    save_page3.click(
        fn=save_structure_page3,
        inputs=my_structure,
        outputs=[
            plan,
            structure_state,
            page3_message
        ]
    )


    essay.change(
        fn=count_words,
        inputs=essay,
        outputs=word_counter
    )


    submit_btn.click(
        fn=submit_essay,
        inputs=essay,
        outputs=submit_message
    )


    generate.click(
        fn=create_final,
        inputs=essay,
        outputs=final_output
    )


# ===================================================
# RUN
# ===================================================

demo.launch(debug=True)
```
