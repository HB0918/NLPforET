```python
# 1. Gradio 설치
!pip install gradio -q

# 2. 라이브러리 불러오기
import gradio as gr
import random

# 3. 문제 데이터
questions = [
    {
        "sentence": "She is a student.",
        "answer": "1",
        "explanation": "'is'는 'She = a student'로, 정체를 설명하므로 '~이다'입니다."
    },
    {
        "sentence": "He is happy.",
        "answer": "2",
        "explanation": "'happy'는 상태를 나타내므로 '~하다(상태)'입니다."
    },
    {
        "sentence": "The book is on the table.",
        "answer": "3",
        "explanation": "위치를 나타내므로 '~에 있다'입니다."
    },
    {
        "sentence": "They are doctors.",
        "answer": "1",
        "explanation": "'They = doctors'로 직업을 설명하므로 '~이다'입니다."
    },
    {
        "sentence": "She is tired.",
        "answer": "2",
        "explanation": "'tired'는 상태를 나타내므로 '~하다(상태)'입니다."
    },
    {
        "sentence": "The cat is under the chair.",
        "answer": "3",
        "explanation": "위치를 나타내므로 '~에 있다'입니다."
    }
]

# 4. 상태 변수
current_question = {}
score = 0
total = 0

# 5. 문제 불러오기
def load_question():
    global current_question
    current_question = random.choice(questions)
    return current_question["sentence"], "", f"점수: {score} / {total}"

# 6. 정답 체크
def check_answer(choice):
    global score, total

    if not current_question:
        return "⚠️ 문제를 먼저 불러오세요!", f"점수: {score} / {total}"

    total += 1

    if choice == current_question["answer"]:
        score += 1
        result = "✅ 정답입니다!"
    else:
        result = f"❌ 틀렸습니다.\n\n해설: {current_question['explanation']}"

    return result, f"점수: {score} / {total}"

# 7. UI 구성
with gr.Blocks() as app:
    gr.Markdown("# 📘 be동사 의미 구분 학습 앱")
    gr.Markdown("문장을 보고 be동사의 의미를 선택하세요.")

    sentence = gr.Textbox(label="문장", interactive=False)

    with gr.Row():
        btn1 = gr.Button("1. ~이다")
        btn2 = gr.Button("2. ~하다")
        btn3 = gr.Button("3. ~에 있다")

    output = gr.Textbox(label="결과")
    score_box = gr.Textbox(label="점수", interactive=False)

    load_btn = gr.Button("🔄 문제 불러오기")

    # 이벤트 연결
    load_btn.click(load_question, outputs=[sentence, output, score_box])

    btn1.click(lambda: check_answer("1"), outputs=[output, score_box])
    btn2.click(lambda: check_answer("2"), outputs=[output, score_box])
    btn3.click(lambda: check_answer("3"), outputs=[output, score_box])

# 8. 실행
app.launch(share=True)
```
